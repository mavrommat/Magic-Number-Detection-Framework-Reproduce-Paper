```python
import sys
#!{sys.executable} -m pip install -r requirements.txt
from astropy.io import ascii
import numpy as np
import pandas as pd
import pprint

# Import the MagicNumberDetector class from the get_magic_numbers module
from Magic_Number_Detection.get_magic_numbers import MagicNumberDetector
import logging

logging.getLogger("matplotlib").setLevel(logging.WARNING)
```

```python
# ReadMe and Data URLs for the CDS dataset
readme_url = "https://cdsarc.cds.unistra.fr/ftp/J/A+A/531/A6/ReadMe"
data_url = "https://cdsarc.cds.unistra.fr/ftp/J/A+A/531/A6/tsi_hol.dat"

# Read the table using CDS format
table = ascii.read(data_url, readme=readme_url, format="cds")
df = table.to_pandas()
print(df[:9])

# Initialize the MagicNumberDetector with desired parameters
detector = MagicNumberDetector(
    sign_violation_threshold=3,
    gauss_threshold=0.1,
    overlap_threshold = 0.1,
    plot_graphs=False, # Set to True to visualize the KDEs plots
    min_unique_values=20
)
# Run the magic number detection
master_results, cleaned_results = detector.run_magic_detection(df)
print("--- Cleaned Magic Number Results ---")
pprint.pprint(cleaned_results)
```
